```python
import os
import shutil

def setup_and_run_pipeline():
    # Cập nhật đường dẫn mới theo dataset bạn vừa up
    input_dir = '/kaggle/input/datasets/laxurie/mouis-final/MODEL_Final'
    working_dir = '/kaggle/working/MODEL_Final'

    # 1. Copy dataset to writable working directory
    if not os.path.exists(working_dir):
        shutil.copytree(input_dir, working_dir)

    # 2. Set working directory
    os.chdir(working_dir)

    # 3. Install required dependencies
    os.system('pip install -q catboost lightgbm xgboost')

    # 4. Execute the modeling pipeline
    print("Executing modeling pipeline...")
    os.system('python pipeline.py')

    # 5. Export final submission file
    # Đã cập nhật tên file theo bản clean mới nhất
    champion_file = "final_submission.csv"
    submission_path = "/kaggle/working/submission.csv"
    
    if os.path.exists(champion_file):
        shutil.copy(champion_file, submission_path)
        print(f"Success: Submission file saved to {submission_path}")
    else:
        print("Error: Pipeline failed to generate the champion submission file.")

if __name__ == "__main__":
    setup_and_run_pipeline()
```
